## Topic 14: Fit Preprocessing Only on Training Data

### Question / Task

### Python Code

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
X = pd.DataFrame({"Income": [20, 30, 40, 50, 60, 70]})
X_train, X_test = train_test_split(X, test_size=0.33, random_state=42)
scaler = StandardScaler().fit(X_train)
print("Train:", scaler.transform(X_train).round(2))
print("Test:", scaler.transform(X_test).round(2))
```

## Topic 17: Detect and Treat Outliers

### Question / Task
Detect outliers using IQR, then cap them instead of automatically removing rows.

### Python Code

```python
s = pd.Series([10, 12, 13, 14, 15, 100])
q1, q3 = s.quantile([0.25, 0.75]); iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print("Outliers:", s[(s < low) | (s > high)].tolist())
print("Capped:", s.clip(low, high).tolist())
```

## Topic 19: Build a Complete Preprocessing Pipeline

### Question / Task
Build a complete pipeline that imputes and scales numerical data and imputes and one-hot encodes categorical data. Split first, fit only on training data.

### Python Code

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.DataFrame({
    "Age": [22, 25, np.nan, 30, 35, 40],
    "City": ["Bangalore", "Mysore", None, "Hassan", "Bangalore", "Mysore"]
})

X = df.replace({None: np.nan})

X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)

num = Pipeline([
    ("fill", SimpleImputer(strategy="mean")),
    ("scale", StandardScaler())
])

cat = Pipeline([
    ("fill", SimpleImputer(strategy="most_frequent")),
    ("encode", OneHotEncoder(handle_unknown="ignore"))
])

preprocess = ColumnTransformer([
    ("num", num, ["Age"]),
    ("cat", cat, ["City"])
])

print("Train shape:", preprocess.fit_transform(X_train).shape)
print("Test shape:", preprocess.transform(X_test).shape)
```
