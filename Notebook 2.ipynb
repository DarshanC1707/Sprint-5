# Notebook 2: Data Type Handling


```python
import pandas as pd
import numpy as np
```

## 1. Identifying Data Types

### Definition

Identifying data types means checking how each column is stored in a dataset. Pandas can distinguish integers, decimals, text, booleans, and dates. This inspection reveals whether values have been imported correctly. It is an important first step before cleaning data.

### Business Example

A shop records prices, product names, and delivery dates. The analyst checks each column before calculating revenue. A price stored as text cannot be summed reliably. Finding this early prevents incorrect reports.

### AI/ML Use

ML preprocessing depends on feature types. Numeric values may need scaling, while categories may need encoding. Date columns may be used to create useful time features. Identifying types helps select the correct transformation.

### Python Example

```python
df = pd.DataFrame({"Age": [22, 25], "City": ["Hassan", "Bengaluru"], "Salary": [25000.0, 30000.0]})
print(df.dtypes)
```

### Explanation

The DataFrame contains age, city, and salary. The dtypes attribute displays how Pandas stores each column. Age is an integer, City is text, and Salary is a float. This helps us choose suitable preprocessing steps.

## 2. Numerical Data

### Definition

Numerical data contains values that represent counts or measurements. Examples include age, salary, and distance. It can be integer or decimal data. Numerical columns support arithmetic and statistical calculations.

### Business Example

A business tracks monthly sales and the number of orders. These are numerical measurements. Managers calculate totals and averages from them. The results help evaluate performance.

### AI/ML Use

Models can use numerical features to learn relationships with targets. Such features are often scaled when necessary. Missing numerical values may be filled using the mean or median. Correct numeric storage is essential for these operations.

### Python Example

```python
df = pd.DataFrame({"Age": [22, 25, 30]})
print(df["Age"].mean())
```

### Explanation

Age is stored as a numerical column. The mean method calculates its average. Pandas can perform arithmetic because the values are numbers. This is useful during data exploration.

## 3. Categorical Data

### Definition

Categorical data describes groups or labels rather than measurements. Examples include city, department, and product type. Categories may be stored as strings or the category dtype. They are usually summarized using counts.

### Business Example

An online store groups orders by payment method. Customers may pay by cash, card, or UPI. Counting each category shows customer preferences. The store can use this information for planning.

### AI/ML Use

Many ML models require categorical labels to be encoded as numbers. One-hot encoding is useful for categories without order. Ordinal encoding may suit genuinely ordered categories. Detecting categorical columns helps prevent incorrect scaling.

### Python Example

```python
df = pd.DataFrame({"City": ["Hassan", "Mysuru", "Hassan"]})
print(df["City"].value_counts())
```

### Explanation

The City column contains category labels. value_counts counts each unique city. Hassan appears twice in this sample. The output shows the category distribution.

## 4. Boolean Data

### Definition

Boolean data represents two logical states: True and False. It is useful for yes/no conditions. Pandas supports the bool dtype. Boolean columns can also be used to filter rows.

### Business Example

A company records whether an order was delivered. Each order is marked True or False. Managers can count delivered orders. This helps track delivery completion.

### AI/ML Use

Boolean values can represent useful binary features. They can indicate membership, status, or conditions. Models may accept booleans or convert them to zero and one. The meaning of each state must remain consistent.

### Python Example

```python
df = pd.DataFrame({"Delivered": [True, False, True]})
print(df["Delivered"].sum())
```

### Explanation

Delivered contains True and False values. When summed, True counts as one and False as zero. The result is two delivered orders. This is a simple way to count positive conditions.

## 5. Date/Time Data

### Definition

Date/time data represents dates, times, or timestamps. Examples include order dates and login times. Pandas uses datetime types for many time operations. Correct conversion allows extraction of day, month, and year.

### Business Example

A retailer records the date of each purchase. Managers compare orders across months. Text dates must first be converted to proper dates. This makes time-based reporting easier.

### AI/ML Use

Dates can provide features such as month, weekday, or elapsed time. These may help models learn seasonal patterns. Incorrect date formats can cause errors. Date features should be created without leaking future information.

### Python Example

```python
df = pd.DataFrame({"Order_Date": pd.to_datetime(["2026-01-10", "2026-02-15"])})
print(df["Order_Date"].dt.month)
```

### Explanation

Order_Date is converted into datetime values. The dt.month accessor extracts month numbers. The results are 1 and 2. These values can be used for monthly analysis.

## 6. String Data

### Definition

String data contains text such as names, addresses, or product codes. Pandas may store text as object or string dtype. Text can include spaces and inconsistent capitalization. Cleaning text improves consistency.

### Business Example

A company collects customer names from registration forms. Some names contain extra spaces. Others use inconsistent capitalization. String operations help standardize these entries.

### AI/ML Use

Text features often need special preprocessing. Simple labels may be encoded, while free text may need vectorization. Cleaning spaces and case can reduce unnecessary differences. Identifiers should not automatically be treated as predictive features.

### Python Example

```python
df = pd.DataFrame({"City": [" hassan ", " BENGALURU "]})
print(df["City"].str.strip().str.title())
```

### Explanation

The City values contain extra spaces and mixed capitalization. strip removes spaces at both ends. title converts words to title case. The cleaned text is more consistent.

## 7. Converting Data Types

### Definition

Converting data types changes how values are stored. It may turn text into numbers, dates, or categories. Conversions must match the meaning of the values. Invalid inputs should be checked rather than silently ignored.

### Business Example

A payroll file stores salaries as text. The analyst converts them to numeric values. This allows average and total salary calculations. Incorrect entries must be identified before reporting.

### AI/ML Use

Correct types make feature engineering and preprocessing possible. Numerical scaling requires numeric inputs. Date extraction requires datetime values. Reliable conversion reduces training errors.

### Python Example

```python
df = pd.DataFrame({"Salary": ["25000", "30000"]})
df["Salary"] = df["Salary"].astype(int)
print(df.dtypes)
```

### Explanation

Salary begins as text. astype(int) converts valid values to integers. dtypes confirms the new storage type. Numeric calculations can now be performed.

## 8. astype()

### Definition

astype() converts a Pandas Series or DataFrame to a specified dtype. It can change integers to floats or text to categories. The conversion works when the values are compatible. Invalid values may raise an error.

### Business Example

A store wants product IDs treated as labels. It can convert the ID column to text. A separate price column can be converted to float. Choosing types helps avoid unintended calculations.

### AI/ML Use

astype is commonly used before feature transformations. Category columns can be marked explicitly. Numeric columns can be converted to a consistent precision. It should be used after checking the data values.

### Python Example

```python
df = pd.DataFrame({"Age": [20, 25]})
df["Age"] = df["Age"].astype(float)
print(df.dtypes)
```

### Explanation

Age originally contains integers. astype(float) changes it to a decimal numeric type. The values become suitable for calculations involving decimals. The output confirms the conversion.

## 9. to_numeric()

### Definition

to_numeric() converts suitable text values into numbers. It supports integer and decimal inputs. The errors argument controls invalid entries. Using errors="coerce" turns invalid entries into NaN.

### Business Example

A sales file contains amounts entered as text. Some entries contain typing mistakes. to_numeric converts valid amounts and flags invalid ones. The business can review the missing results.

### AI/ML Use

ML algorithms generally require numeric inputs for numeric features. to_numeric prepares imported text values for calculations. Coerced invalid entries can then be handled as missing data. This avoids treating invalid text as valid measurements.

### Python Example

```python
df = pd.DataFrame({"Price": ["100", "250", "bad"]})
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
print(df)
```

### Explanation

The two valid prices become numeric values. The word bad cannot be converted. errors="coerce" changes it to NaN. The invalid entry can now be investigated.

## 10. to_datetime()

### Definition

to_datetime() converts suitable values into Pandas dates or timestamps. It accepts common date formats. A specified format can help avoid ambiguity. Invalid values can be converted to NaT for review.

### Business Example

An office imports joining dates from a spreadsheet. Dates initially appear as strings. The analyst converts them to datetime values. This makes it possible to calculate employment duration.

### AI/ML Use

Date conversion enables time-based feature engineering. Models may use month or day-of-week features. Invalid dates must be detected and handled. Proper timestamps also support chronological train/test splitting.

### Python Example

```python
df = pd.DataFrame({"Date": ["2026-01-05", "2026-02-10"]})
df["Date"] = pd.to_datetime(df["Date"])
print(df["Date"].dt.year)
```

### Explanation

The strings are converted into datetime values. The dt.year accessor extracts the year. Both rows return 2026. This demonstrates a simple date feature.

## 11. Handling Invalid Conversions

### Definition

Invalid conversions occur when values do not match the requested data type. Examples include letters inside numeric fields or impossible dates. Pandas may raise an error or mark invalid values as missing. These cases should be reviewed before modeling.

### Business Example

A shop imports quantities with an entry like unknown. Direct integer conversion fails on that entry. The analyst marks it as missing and investigates it. This prevents inaccurate stock calculations.

### AI/ML Use

Invalid values can break preprocessing pipelines. Converting them to missing values allows deliberate imputation or exclusion. The chosen approach depends on the business meaning. Silent removal may hide important data quality problems.

### Python Example

```python
df = pd.DataFrame({"Quantity": ["5", "wrong", "8"]})
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
print(df["Quantity"].isna().sum())
```

### Explanation

The word wrong cannot become a number. to_numeric marks that value as NaN. isna().sum() counts one invalid entry. The dataset can now be corrected intentionally.

## 12. Detecting Incorrect Data Types

### Definition

Incorrect data types occur when columns are stored in a format that does not match their meaning. Numeric-looking text is a common example. Checking dtypes and sample values helps reveal problems. Type validation should happen early in preprocessing.

### Business Example

A bank receives an account balance column as text. This prevents direct numeric calculations. The analyst inspects the dtype and a few values. The column can then be safely converted.

### AI/ML Use

Incorrect types may stop model training or cause misleading transformations. Inspecting data types is part of data validation. Numeric, categorical, and date features need different preparation. Detecting problems early keeps the workflow reliable.

### Python Example

```python
df = pd.DataFrame({"Balance": ["1000", "2000"]})
print(df.dtypes)
print(pd.to_numeric(df["Balance"]).dtype)
```

### Explanation

Balance is first stored as object because its values are strings. to_numeric produces a numeric Series. Comparing the two dtypes reveals the issue. The converted values can then be saved back into the column.

## 13. String → Numeric

### Definition

String-to-numeric conversion changes text containing digits into numeric values. It is needed when numbers are imported as text. Pandas provides to_numeric for this task. Invalid strings should be checked.

### Business Example

A store receives product prices as strings. The prices must be numeric to calculate revenue. Conversion allows totals and averages. Invalid price entries can be flagged.

### AI/ML Use

Numeric features must be correctly typed before scaling or modeling. String values may cause estimators to fail. Conversion helps ensure consistent input. Invalid values can be handled separately.

### Python Example

```python
df = pd.DataFrame({"Price": ["120", "250"]})
df["Price"] = pd.to_numeric(df["Price"])
print(df.dtypes)
```

### Explanation

Price starts as a text column. to_numeric converts the values into numbers. The dtype output confirms the result. Arithmetic is now possible.

## 14. String → Date

### Definition

String-to-date conversion changes date text into datetime values. Date strings may look readable but lack datetime functionality. to_datetime performs this conversion. Explicit formats are useful for ambiguous dates.

### Business Example

A business stores invoice dates as strings. Analysts need to group invoices by month. Converting them to datetime makes that possible. Invalid dates can be reviewed.

### AI/ML Use

Datetime features support seasonal and chronological analysis. Models can use extracted month or weekday values. Dates should be parsed consistently. Time ordering also matters for avoiding leakage.

### Python Example

```python
df = pd.DataFrame({"Date": ["2026-03-01", "2026-03-02"]})
df["Date"] = pd.to_datetime(df["Date"])
print(df["Date"].dt.day)
```

### Explanation

The Date column begins as text. to_datetime converts it to a datetime dtype. dt.day extracts the day of the month. The output contains 1 and 2.

## 15. Numeric → Categorical

### Definition

Numeric-to-categorical conversion treats numeric codes as labels. Some numbers represent categories rather than quantities. For example, a department code may not have mathematical meaning. Pandas supports the category dtype.

### Business Example

A company uses department codes 1, 2, and 3. These codes identify teams rather than amounts. Treating them as categories avoids misleading averages. Reports can count employees by department.

### AI/ML Use

ML preprocessing should distinguish numeric measurements from coded categories. Coded labels often require categorical encoding. Treating them as continuous numbers may imply a false order. Correct typing helps select appropriate encoders.

### Python Example

```python
df = pd.DataFrame({"Department": [1, 2, 1, 3]})
df["Department"] = df["Department"].astype("category")
print(df.dtypes)
```

### Explanation

Department contains numeric labels. astype("category") marks them as categories. The values remain the same. Pandas now recognizes their categorical meaning.

## 16. Integer → Float

### Definition

Integer-to-float conversion changes whole-number storage to decimal-number storage. Float values can represent fractional results. Pandas supports this through astype(float). It is useful when calculations require decimals.

### Business Example

A shop initially stores unit prices as whole numbers. Discounts may create fractional prices. Converting the column to float supports those results. The amounts can then be calculated accurately within normal floating-point limits.

### AI/ML Use

Many numerical transformations produce decimal values. Scaling and imputation may turn integers into floats. A consistent numeric type can simplify processing. Floating-point rounding should be considered for exact financial calculations.

### Python Example

```python
df = pd.DataFrame({"Price": [100, 200]})
df["Price"] = df["Price"].astype(float)
print(df)
```

### Explanation

The original prices are integers. astype(float) converts them into decimal values. The output displays 100.0 and 200.0. The column can now store fractional prices.

## 17. Why Correct Data Types Are Important

### Definition

Correct data types ensure that values are interpreted properly. Numbers should support arithmetic, dates should support time operations, and categories should represent labels. Wrong types can cause errors or misleading results. Checking and correcting them improves data quality.

### Business Example

A company cannot accurately calculate average sales if amounts are stored as text. It also cannot easily group transactions by month if dates are strings. Correcting both columns makes reporting easier. This supports better business decisions.

### AI/ML Use

ML workflows depend on data types to choose preprocessing methods. Numeric columns may need scaling, while categorical columns need encoding. Correct dates support time features and time-aware splits. Type validation helps make model inputs consistent and reliable.

### Python Example

```python
df = pd.DataFrame({"Sales": ["100", "200"]})
df["Sales"] = pd.to_numeric(df["Sales"])
print("Average sales:", df["Sales"].mean())
```

### Explanation

Sales begins as strings. to_numeric converts the column into numbers. The mean method then calculates the average as 150.0. Correct typing makes the analysis possible.
